### Split Dream Journal into .txt files
- Breaks entire journal into individual dreams (based on date), placed into folders for year
- Based on format of my dream journal:
>MM/DD/YY
>
>#tag1 #tag2
>
>Content of first dream...
>
>Content of second dream...
- If the format of your dream journal differs, the code will need to be changed here

```python
cont = content.split('\n') # Split journal based on line breaks
date_pattern = re.compile(r'^\d{1,2}/\d{1,2}/\d{2}$') # format of dates

final = []
current = []
for line in cont: # Goes through each line, if it's a date, add it and everything up until next date to a sub-list
    if date_pattern.match(line):
        if current:
            final.append(current)
        current = [line]
    else:
        current.append(line)
if current:
    final.append(current)

print(final)
```

```python
n = 1
count = 0
for dream in final:
    date = dream[0].split("/") # Break date into month, day, year, for reformatting and organizing
    name = '_'.join(date)
    if date[-1] == "00": # If date was unsure, I used 0s. All unknown were prior to 6/19, so they go in the 2019 folder
        date[-1] = "19"
    try:
        f = open(save_dir + "20" + date[-1] + "/" + name + ".txt", "x", encoding="utf-8")
    except: # Multiple dates are 0/0/00, this catches errors from multiple files having the same name and adds an extra index to end
        f = open(save_dir + "20" + date[-1] + "/" + name + "_" + str(n) + ".txt", "x", encoding="utf-8")
        n += 1
    text = '\n'.join(dream)
    f.write(text)
    f.close()
    print(name + " saved.")
    count += 1
print(count)
```

### Date, Tag, Content Extraction
- extract tags, with their dates
- graph appearance of specific tags over time

```python
dream_dir = "dreams/2018"
year_dreams = []
for filename in os.listdir(dream_dir):
    f = open(dream_dir + '/' + filename, 'r', encoding='utf-8')
    content = f.read()
    f.close()
    cont = content.split('\n')
    date = cont[0]
    file_name = filename
    start_idx = 1
    if cont[1][0] == '#':
        tags = cont[1].split(' ')
        start_idx = 2
    else:
        tags = None

    dream_content = []
    current = []
    for line in cont[start_idx:]: # Goes through each line, if it's a date, add it and everything up until next date to a sub-list
        if line == '':
            if current:
                dream_content.append(' \n'.join(current))
                current = []
        else:
            current.append(line)
    if current:
        dream_content.append(current)
    
    year_dreams.append([file_name, date, tags, dream_content])
```

### LLM Analysis
- Use an LLM to read and analyze dreams to identify prominent themes and topics
- Distinct dreams on the same day each get their own analysis, but should be put into the same file

```python
analysis_save_dir = 'dreams/2018/analysis'
for dream in year_dreams:
    full_reply = []
    for individual_dream in dream[2]: # make it so multiple dreams from same night all get put in same file
        print("Getting analysis for " + str(dream[0]))
        response = ollama.chat(model='deepseek-r1', messages=[
            {
                'role': 'user',
                'content': 'In a few words, identify the most prominent themes in this dream. If there is no clear meaning, do not try to make one up: simply reply "None". Do not rely on universal symbols or interpretations. Do not try to validate or reassure the user. Do not avoid taboo topics. All available information is given.: ' + individual_dream,
            },
        ])
        print("Got analysis for " + str(dream[0]))
        reply = response['message']['content']
        full_reply.append(reply)
    full_themes = []
    for rep in full_reply:
        themes = rep.split('</think>')[-1]
        full_themes.append(themes)
    full_reply = '\n'.join(full_reply)
    full_themes = '\n'.join(full_themes)
    name = dream[0]
    long = open(analysis_save_dir + "/long/" + name + ".txt", "x", encoding="utf-8")
    long.write(full_themes)
    long.close()
    if full_themes != 'None':
        short = open(analysis_save_dir + "/short/" + name + ".txt", "x", encoding="utf-8")
        short.write(full_themes)
        short.close()
    print(name + " saved.")
    
```
